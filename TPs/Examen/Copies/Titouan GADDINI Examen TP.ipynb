```python
#EX 1 3.
def Dirichlet(m,n,K):
    if gcd(m,n)!=1:
        return False
    L=[]
    a=K
    s=0
    while a!=0:
        if is_prime(m+s*n):
            L.append(m+s*n)
            a=a-1
        s+=1
    return L
    
#et Dirichlet(3,100,10) donne [3, 103, 503, 1103, 1303, 2003, 2203, 2503, 2803, 2903]
```

```python
#EX 2 1.
def suite_Un(n):
    u=0
    v=1
    if n==0:
        return u
    if n==1:
        return v
    for i in range(n):
        c=v
        v=v+u
        u=c
    return v

       
```

```python
#EX 2 2.
def Un_mod(n,m):
    u=0
    v=1
    if n==0:
        return u
    if n==1:
        return v
    for i in range(n):
        c=v
        v=(v+u)%m
        u=c
    return v
#et Un_mod(10**7,10) qui vaut le reste de U_10**7 dans la division euclidienne par 10 est 1
```

```python
#EX 3 1.
def premier(n):
    m=floor(sqrt(n))
    if n%2==0:
        return(False,2)
    for i in range (3,m+1,2):
        if n%i==0:
            return(False,i)
    return True
```

```python
#EX 5 1.
def cryptage(cle,texte):
    C=""
    L=len(cle)
    for i in range(len(texte)):
        C+=chr((ord(texte[i])+ord(cle[i%L])-2*65)%26+65)
    return C
```
