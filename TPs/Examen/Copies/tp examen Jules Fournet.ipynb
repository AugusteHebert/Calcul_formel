```python
#4)
def triplets():    #je fais tourner la prochaine fonction sur toutes les valeurs de 0 à 25
    tab=[]
    for i in range(0,26): 
        res=triplets_c(i)
        if res!=(0,0):
            tab.append(res)
    return tab
        
        
def triplets_c(c):         #je touve le triplet pour une certaine valeur c
    for i in range(1,c):              
        for j in range(1,c):
            if (i**2)+(j**2)==c:
                return (i,j)
    return (0,0)
```

```python
#EXO3
#1)
def premier(n):
    for i in range(2,racine(n)+1):
        if n%i==0:
            return (false,i)
    return true

def racine(n):         #calcule le premier entier supérieur à sqrt(n)
    for i in range(1,n):
        if (i**2<=n) and ((i+1)**2>n):
            return i
    return 0

premier(9)
```

```python
#4)
def produits(n):
    res=[]
    L=diviseurs(n)
    for i in range(len(L)):
        j=puissance_premier(n,L[i])
        res.append((L[i],j))
    return res
        
        
def puissance_premier(n,p):    #calcule la puissance de chaque nombre premier dans l'entier n
    i=0
    while n%p==0:
        i+=1
        n=n/p
    return i
```

```python
#EXO5
#1)
def crypter(cle,texte):      
    res=""
    n=len(texte)
    m=len(cle)
    for i in range(0,n):
        j=i%m
        elt=decalage(texte[i],(ord(cle[j])-65))
        res=res+elt
    return res
                
                
def decalage(lettre,i):      #je change une lettre avec la méthode de décalage
    code_ascii=ord(lettre)-65
    nouveau_code=code_ascii-i
    if nouveau_code<0:
        nouveau_code=26+nouveau_code
    return chr(nouveau_code+65)
```

```python
#EXO6
#1)
def premier_aleatoire(k):
    res=1
    while is_prime(res)==false:
        res=ZZ.random_element(10**k,1+10**(k+1))
    return res
```

```python
#2)
def pgcd(x,y):
    if x<y:
        cpt=x
        x=y
        y=cpt
    if y==1:
        return x
    if y==0:
        return x
    return pgcd(y,x%y)

def choix_e(phi):  #donne un element e
    e=phi
    while pgcd(e,phi)!=1:
        e=ZZ.random_element(2,phi)
    return e

def choix_d(phi,e):    #donne un element d
    d=0
    while (e*d)%phi!=1:
        d=ZZ.random_element(2,phi)
    return d
    

def cle(k):
    p=premier_aleatoire(k)
    q=premier_aleatoire(k)
    n=p*q
    phi=(p-1)*(q-1)
    e=choix_e(phi)
    d=choix_d(phi,e)
    return [(n,e),(d,phi)]
```
