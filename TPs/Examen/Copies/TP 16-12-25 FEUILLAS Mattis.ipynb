```python
def factorielle(n):
    if n==0:
        return 1
    res=1
    for i in range(1,n+1):
        res=res*i
    return res
```

```python
def decomposition(n):
    s=0
    while n%(2**s)==0:
        s=s+1
    s=s-1
    return (s,n/(2**s))
```

```python
def dirichlet(m,n,K):
    if gcd(m,n)!=1:
        return 'Erreur'
    L=[]
    s=2
    while len(L)<=K:
        if is_prime(s) and s%n==m:
            L.append(s)
        s=s+1
    return L[:-1]
```

```python
def pythagore(c):
    L=[]
    for i in range(c+1):
        for a in range(0,i+1):
            for b in range(0,i+1):
                if a**2 + b**2 == i**2:
                    L.append((a,b,i))
    return L
```

```python
def u(n):
    if n==0:
        return 0
    if n==1:
        return 1
    L=[0,1]
    for i in range(2,n+1):
        L.append(L[i-1]+L[i-2])
    return L[n]
```

```python
def divpremiers(n):
    L=[]
    for i in range(2,n+1):
        if is_prime(i) and n%i==0:
            L.append(i)
    return L
```

```python
def decfactprem(n):
    M=divpremiers(n)
    L=[]
    for i in range(len(M)):
        s=1
        while n%(M[i]**s)==0:
            s=s+1
        s=s-1
        L.append((M[i],s))
    return L
```

```python
def temoineuler(n,a):
    if is_prime(n):
        return False
    symb=jacobi_symbol(a,n)%n
    puiss=power_mod(a,(n-1)/2,n)
    return symb != puiss
```

```python
def solovay_strassen(n,k):
    for i in range(k):
        a=ZZ.random_element(2,n)
        if temoineuler(n,a):
            return "n n'est pas premier"
    return 'n est peut-être premier'
```

```python
def premier_aleatoire(K):
    s=0
    while s<1:
        p=ZZ.random_element(10**K,10**(K+1) +1)
        if is_prime(p):
            s=s+1
            return p
```

```python
premier_aleatoire(3)
```

```python
#Q2
```

```python
def cle(K):
    #On génère p,q et on calcule n et phi
    p=premier_aleatoire(K)
    q=premier_aleatoire(K)
    n=p*q
    phi=(p-1)*(q-1)
    #On génère e premier avec phi
    s=0
    e=None
    while s<1:
        a=ZZ.random_element(2,phi)
        if gcd(a,phi)==1:
            e=a
            s=s+1
    #Calcul de d
    (u,v,w)=xgcd(e,phi) #u est égal à 1, c'est le pgcd... et v*e + w*phi = u = 1
    d = v%phi
    return [(n,e),(d,phi)]
```

```python
cle(3)
```
