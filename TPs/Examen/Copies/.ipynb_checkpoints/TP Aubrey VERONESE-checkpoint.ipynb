```python
#Exercice 1
def factorielle(n):
    x=1
    for i in range(1,n+1):
        x*=i
    return x

factorielle(1000)
```

```python
def decomposition(n):
    t=n
    s=0
    while t%2==0:
        s+=1
        t=t/2
    return s,t

decomposition(20)
```

```python
def fo(m,n,K):
    if gcd(m,n)>1:
        return "erreur"
    L=[]
    i=m
    while len(L)<K:
        if is_prime(i):
            L.append(i)
        i+=n
    return L

fo(3,100,10)
```

```python
#Exercice 2
D={0 :0, 1:1}
def u(n):
    if n in D:
        return D[n]
    a,b=u(n-1),u(n-2)
    D[n]=a+b
    return a+b
```

```python
#Exercice 3
def f(n):
    i=2
    while i<n**.5:
        if n%i==0:
            return (False,i)
        i+=1
    return True
```

```python
def listepremiers(n):
    L=[2]
    if n==1:
        return L
    i=3
    while len(L)<n:
        if is_prime(i):
            L.append(i)
        i+=2
    return L

listepremiers(1000)[-1]
```

```python
def diviseurspremiers(n):
    p=2
    L=[]
    if is_prime(n):
        return [n]
    while p<n/2:
        if n%p==0:
            L.append(p)
        p=next_prime(p)
    return L

diviseurspremiers(360)
```

```python
def fon(n):
    L=diviseurspremiers(n)
    P=[]
    for p in L:
        alpha=1
        while n%p==0:
            n=n/p
            alpha+=1
        P.append((p,alpha-1))
    return(P)      

fon(360)
```

```python
#ex 5
alphabet1,alphabet2={},{}
for i in range(ord('A'),ord('Z')+1):
    alphabet1[chr(i)]=i-65                       #alphabet1[i] renvoie la i-eme lettre
    alphabet2[i-65]=chr(i)

def cryt(cle,texte):
    m=len(cle)
    textecrypte=''
    for i in range(len(texte)):
        lettre=texte[i]
        decalage=alphabet1[cle[i%m]]
        textecrypte=textecrypte+alphabet2[(decalage+alphabet1[lettre])%26]
    return textecrypte

cryt("python","unexempledechiffre")
```

```python
def decryt(cle,texte):
    m=len(cle)
    textecrypte=''
    for i in range(len(texte)):
        lettre=texte[i]
        decalage=alphabet1[cle[i%m]]
        textecrypte=textecrypte+alphabet2[(-decalage+alphabet1[lettre])%26]
    return textecrypte

```

```python
#exercive 6
def premier_aleatoire(K):
    m=1
    while not(is_prime(m)):
        m=ZZ.random_element(10^K,10^(K+1)+1)
    return m

for i in range(5):
    print(premier_aleatoire(5))
```

```python
def cle(K):
    p,q=0,0
    while p==q:
        p,q=premier_aleatoire(K),premier_aleatoire(K)
    n,phi=p*q,(p-1)*(q-1)
    e=2
    while gcd(e,phi)>1:
        e+=1
    d=2
    while (e*d)%phi!=1:
        d+=1
    return [n,e],[d,phi]
```

```python
cle(3)
```
