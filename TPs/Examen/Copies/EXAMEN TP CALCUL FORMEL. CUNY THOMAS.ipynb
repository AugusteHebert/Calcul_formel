```python
#4
def pythagore(n): #cette fonction détermine les triplets (a,b,c) tels que a^2 + b^2 = c^2 où c au plus n.
    L=[] # liste des triplets
    for a in range(n+1):
        for b in range(n+1):
            for c in range(n+1):
                if a^2 + b^2 == c^2:
                    L.append((a,b,c))
    return L
```

```python
#Exercice 2
#1 
def fibo(n):
    u=0
    v=1
    for k in range(1,n):
        c=u
        u=v
        v=c+u
    return v
```

```python
#4
def decomp(n,p): #cette fonction sert à calculer les valuations p-adiques de n
    s=0
    u=n
    while u%p ==0:
        s=s+1
        u=u/p
    return s


def factoriser(n):
    L = listdiv(n)
    q=len(L)
    L1 = []
    for k in range(q):
        L1.append( (L[k],decomp(n,L[k])) )
    return L1
    
```

```python
#2
#On définit pour alléger un peu le code de cle une autre fonction :
def inverse(u,n): #on suppose u premier à n
    for k in range(1,n):
        if (u*k-1) % n ==0:
            return k
        

def cle(K):
    p,q = premier_aleatoire(K),premier_aleatoire(K)
    phi=(p-1)*(q-1)
    e = phi-1 # on a bien pgcd(e,phi)=1
    d = inverse(e,phi)
    return [(pq,e),(d,phi)]
    
    
```

```python
#5
def decrypte(C,n,e):
    for m in range(1,n):
        a = m^e - C
        if a%n==0:
            return m
```
