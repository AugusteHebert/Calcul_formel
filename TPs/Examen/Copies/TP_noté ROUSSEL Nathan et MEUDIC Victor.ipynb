```python
def decomposition (n):
    s=0
    t=n
    if t%2==1:
        return (s,t)
    while t%2==0:
        s+=1
        t=n/2
    return (s,t)
```

```python
def triplet_pythagoricien (n):
    l=[0]
    for c in range (n):
        for a in range (c):
            for b in range (c):
                if a**2+b**2==c**2:
                    l.append((a,b,c))
    return l
```

```python
def fib(n):
    L=[0,1,1,2]
    if n<4:
        return L[n]
    for i in range (4,n):
        L.append(L[i-1]+L[i-2])
    return L[n-1]
```

```python
def fib_mod_10 (n):
    L=0
    L_suiv=1
    if n==0:
        return 0
    if n ==1:
        return 1
    for i in range (2,n):
        temp=L_suiv
        L_suiv=((L+L_suiv)%10)
        L=temp
    return L_suiv
```

```python
print(fib_mod_10(10**7))
```

```python
def x_prime( n):
    i=0
    j=0
    while i<=n:
        j+=1
        if is_prime(j) :
            i+=1
    return j
```

```python
def crypter (cle,texte):
    res=""
    taille=len (cle)
    for i in range (len (texte)):
        c=cle[i%taille]
        o=ord(c)-64
        caractère_temporaire= texte[i]
        t=ord(caractère_temporaire)
        res=res+chr(t+o)
    return res
```

```python
def premier_aleatoire (K):
    a=randint(10**K,10**(K+1))
    while not is_prime(a):
        a=randint(10**K,10**(K+1))
    return a
```
