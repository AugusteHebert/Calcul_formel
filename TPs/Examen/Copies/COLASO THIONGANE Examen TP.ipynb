```python
#Exercice 1
#Question 2
def decomposition(n):
    if n%2==1:
        return (0,n)
    else:
        q=n//2
        s=0
        while q%2==0:
            q=q//2
            s+=1
        return (s+1,q)
```

```python
#Exercice 1
#Question 3
def Dirichlet(m,n,K):
    if gcd(m,n)!=1:
        return "erreur"
    else:
        L=[]
        c=0
        while K>0:
            if c%n==m:
                L.append(c)
                K-=1
            c+=1
        return L
```

```python
#Exercice 1
#Question 4
def pythagore(n):
    L=[]
    for a in range(n+1):
        for b in range(n+1):
            for c in range(n+1):
                if a^2+b^2==c^2:
                    L.append((a,b,c))
    return L
```

```python
#Liste des triplets pythagoriciens tels que c<=25
print(pythagore(25))
```

```python
#Exercice 3
#Question 1
def premier(n):
    c=floor(sqrt(n))
    for i in range(2,c+1):
        if n%i==0:
            return False,i
    return True
```

```python
#Exercice 3
#Question 3
def diviseurs_premiers(n):
    L=[]
    for i in range(n+1):
        if is_prime(i):   #si i est premier
            if n%i==0:   #si i est un diviseur de n
                L.append(i)
    return L
```

```python
#Exercice 3
#Question 4
def factorisation(n):
    L=[]
    a=0 #initialisation du compteur de puissance
    for i in range(n+1):
        if is_prime(i):   #si i est premier
            if n%i==0:   #si i est un diviseur de n
                while n%(i^a)==0:  # détermination de la valuation i-adique
                    a+=1
                L.append((i,a-1))
                a=0                     #remise à zero du compteur de valuation
    return L
```
