```python
#Exercice 1:
#4

def listTrippletPytha():
    res = []
    for c in range(26):
        c2 = c^2
        for b in range(26):
            b2 = b^2
            for a in range(26):
                if a^2 + b2 == c2:
                    res.append((a, b, c))
    return res
```

```python
#Exercice 3 :
#2
def niemePrime(n):
    compteur = 0
    testedNumber = 0
    while compteur < n:
        if is_prime(testedNumber):
            compteur += 1
        testedNumber += 1
    return testedNumber - 1 
```

```python
#Exercice 5
#1
def convertTextToIndiceListe(text):
    res = []
    for k in text:
        res.append(ord(k) - 65)
    return res

def convertIndiceListeText(liste):
    text = ''
    for k in liste:
        text = text + chr(k+65)
    return text
    
    
def chiffrementVigenere(cle, text):
    listeCle = convertTextToIndiceListe(cle)
    listeText = convertTextToIndiceListe(text)
    compteurCle = 0
    lenCle = len(cle)
    listeCodedIndice = []
    for k in listeText:
        listeCodedIndice.append((k + listeCle[compteurCle] )% 26)
        compteurCle = (compteurCle + 1) % lenCle
    return convertIndiceListeText(listeCodedIndice)
```
