```python
def premier(p):
    for i in range(2,p):
        if p%i==0:
            return False
    return True
        


def dirichlet(m,n,K):
    x=0
    L=[]
    if gcd(m,n)!=1:
        return "Erreur"
    else:
        while K!=0:
            while not premier(x):
                x=x+1
            if x%n==m:
                L.append(x)
                K=K-1
            x=x+1
    return L
            
    
```

```python
dirichlet(5,7,6)
```

```python
def triplet(a,b,c):
    return a**2+b**2==c**2

def tripletjusque_c(c):
    L=[]
    for k in range(c):
        for i in range(k):
            for j in range(k):
                if triplet(i,j,k):
                    L.append((i,j,k))
    return L
```

```python
def diviseur(n):
    L=[]
    if is_prime(n):
        return [n]
    else:
        while not is_prime(n):
            L.append(premier_k(n)[1])
            n=n//premier_k(n)[1]
    return L + [n]

def distinct(L):
    V=[L[0]]
    for i in L:
        if i not in V:
            V.append(i)
    return V

def diviseur_distinct(n):
    return distinct(diviseur(n))
```

```python
def temoin_euler(n,a):
    x=jacobi_symbol(a,n)
    y=a**((n-1)//2)
    
    return not x%n==y%n
```

```python
def Solovay_Strassen(n,k):
    for i in range(k):
        a=randint(2,n-1)
        if temoin_euler(n,a):
            return "Ce nombre n'es pas premier"
    return "Peut-être que ce nombre est premier"
```

```python
def liste_de_peutetre_premier(M,r):
    n=3
    L=[]
    while n<M:
        if Solovay_Strassen(n,r)=="Peut-être que ce nombre est premier":
            L.append(n)
        n=n+2
    return L
```

```python
def mtaille(cle,texte):
    s=""
    for i in range(len(texte)):
        s=s+cle[i%len(cle)]
    return s
    


def chiffrement(cle,texte):
    vrai_cle=mtaille(cle,texte)
    decallage=[ord(vrai_cle[i]) for i in range(len(vrai_cle))]
    
    texteC=""
    for i in range(len(texte)):
        texteC=texteC+chr((ord(texte[i])+decallage[i])%26+65)
    return texteC
```

```python
chiffrement("PYTHON","UNEXEMPLEDECHIFFRE")
```

```python
def decrypter(cle,texte):
    vrai_cle=mtaille(cle,texte)
    decallage=[ord(vrai_cle[i]) for i in range(len(vrai_cle))]
    
    texteC=""
    for i in range(len(texte)):
        texteC=texteC+chr((ord(texte[i])-decallage[i])%26+65)
    return texteC
```

```python
chiffrement("PYTHON",decrypter("PYTHON","UNEXEMPLEDECHIFFRE"))
```

```python
decrypter("PYTHON",chiffrement("PYTHON","UNEXEMPLEDECHIFFRE"))
```

```python
def premier_k(p):
    for i in range(2,floor(sqrt(p))+1):
        if p%i==0:
            return False,i
    return True


def decrypter(C,n,e):
    p=premier_k(n)[1]
    q=n//premier_k(n)[1]
    
    phi=(p-1)*(q-1)
    for i in range(2,phi-1):
        if i*e%phi==1:
            d=i
    return (C**d)%phi
```
