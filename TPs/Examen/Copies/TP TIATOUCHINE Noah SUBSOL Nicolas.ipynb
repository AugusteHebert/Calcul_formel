```python
#Exercice 3 #Q1
def divisibilite(n):
    for k in range (2,int(sqrt(n))):
        if n%k==0:
            return(False,k)
    return True
```

```python
#Q2
def n_premier(n):  
    p=2
    c=1
    while c<n:
        p=p+1
        while not is_prime(p):
            p=p+1
        
        c=c+1
    return p
```

```python
#Q2
n_premier(1000)
```

```python
#Q3
def diviseurs_premiers(n):
    L=[]
    p=2
    while p<=n:
        if n%p==0:
            L.append(p)
        p=next_prime(p)
    return L
            
        
```

```python
diviseurs_premiers(78)
```

```python
#Q4
def decomposition_2(n):
    K=[]
    for p in diviseurs_premiers(n):
        c=0
        m=n
        while m%p==0:
            c=c+1
            m=m//p
        K.append((p,c))
    return K
```

```python
#Q4
decomposition_2(6)
```

```python
#Exercice 4
#Q1
def temoin_(n,a):
    #n impair
    if jacobi_symbol(a,n)!=power_mod(a,(n-1)/2,n):
        return True
    return False
```

```python
#Q2
def testSolovayStrassen(n,k):
    c=0
    while c<k:
        a=ZZ.random_element(2,n)
        if temoin_(n,a):
            return False
        c=c+1
    return True
```

```python
#Exercice 5 #Q1
def Vigenere(cle,texte):
    m=""
    L=[ord(cle[i%len(cle)]) for i in range(len(texte))]
    for k in range(len(texte)):
        e=ord(texte[k])
        e=e-65
        e=(e+(L[k]-65))%26
        e+=65 
        m+=chr(e)
    return m
```

```python
#Q2

def Decrypte(cle,texte):
    m=""
    L=[ord(cle[i%len(cle)]) for i in range(len(texte))]
    for k in range(len(texte)):
        e=ord(texte[k])
        e=e-65
        e=(e-(L[k]-65))%26
        e+=65 
        m+=chr(e)
    return m
```

```python
#Exercice 6
#Q1
def premier_aleatoire(K):
    p=10**(K+1)+1
    while p>10**(K+1):
        p=ZZ.random_element(10**K,10**(K+1)+1)
        p=next_prime(p)
    return p
```
