```python
# Exercice 5
# Question 1
def Vigenère(texte,cle):
    M = ""
    n = len(cle)
    for i in range(len(texte)):
        a = i%n
        o = (ord(cle[a]) -65 + ord(texte[i]) -65)%25 +65
        M += chr(o)
    return M
```
