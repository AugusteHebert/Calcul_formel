```python
def suite(n):
    if n==0:
        return 0
    elif n==1:
        return 1
    else:
        return suite(n-1)+suite(n-2)
```

```python
def chiffrement(cle,texte):
    M=""
    n=len(texte)
    m=len(cle)
    q=n//m
    r=n%m
    for i in range(q):
        for j in range(m):
            a=ord(texte[m*i+j])
            b=ord(cle[j])-65
            c=chr(a+b)
            M+=c
    for k in range(r):
            a=ord(texte[m*q+k])
            b=ord(cle[k])-65
            c=chr((a+b))
            M+=c   
    return M
```

```python
def premier_aleatoire(k):
    x=ZZ.random_element(10^k,10^(k+1))
    while not is_prime(x):
        x=ZZ.random_element(10^k,10^(k+1))
    return x
```

```python
def cle(k):
    p=premier_aleatoire(k)
    q=premier_aleatoire(k)
    n=p*q
    l=(p-1)*(q-1)
    e=2
    while gcd(e,l)!=1:
        e+=1
    d=pow(e,-1,l)
    return [(n,e)],[(d,l)]
```

```python
def crypt(m,p,q,e):
    n=p*q
    c=m
    i=1
    while i <=e:
        c=pow(c,i,n)
        i+=1
    return c
```
