```python
#Ex 5: 1
def vigenere(cle,texte):
    long_t = len(texte)
    long_c = len(cle)
    T = code_ascii(texte)
    C = code_ascii(cle)
    Codage = ''
    for i in range(long_t):
        j = i%long_c
        c= T[i]+ C[j]-65
        if c > 90 :
            c-=26
        Codage += chr(c)
    return Codage
```
