```python
#1
def factorielle(n):
    res = 1
    while n>1 :
        res = res*n
        n = n-1
    return res
```

```python
#2
def decomposition(n):
    s = 1
    if is_prime(n):
        return(0,n)
    if n%2 == 0 :
        t = n//2
        while t%2 == 0 :
            t = t//2
            s = s+1
    if n%2 == 1 :
        s = 0
        t = n        
    return (s,t)
```

```python
#3
def divpremier(n):
    L = []
    a = n
    while a>1:
        if is_prime(a):
            if n%a == 0:
                L = L.append(a)
        a = a-1
    return L
```
