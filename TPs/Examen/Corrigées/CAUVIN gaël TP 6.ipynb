```python
def premier_aleatoire(K):
    p=ZZ.random_element(10**K,10**(K+1))
    while(not is_prime(p)):
        p=ZZ.random_element(10**K,10**(K+1))
    return p
```

```python
def cle(K):
    p=premier_aleatoire(K)
    q=premier_aleatoire(K)
    phi=(p-1)*(q-1)
    n=p*q
    for e in range(2,phi):
        for d in range(2,phi):
            if e*d%phi ==1 :
                t=0
                for i in diviseur(e):
                    for j in diviseur(phi):
                        if i==j:
                            t=-1
                        if t==-1:
                            break
                    if t==-1:
                        break

                if t==-1:
                    break
                return [(n,e),(d,phi)]
```
