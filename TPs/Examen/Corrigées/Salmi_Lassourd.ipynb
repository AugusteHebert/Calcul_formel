```python
def nemeprime(n):
    a=1
    b=0
    while b<n :
        a+=1
        if a.is_prime() :
            b+=1
    return a
```

```python
def liste_fact(n):
    L=[]
    p=2
    a=n
    while p<n:
        if n%p == 0:
            b=1
            a=a//p
            while a%p ==0 :
                b=b+1
                a=a//p
            L.append((p,b))
        p.next_prime()
    return(L)
```

```python
#Exercice 5
def cesar(T,n):
    C = ''
    for l in T:
        ordre = ((ord(l)-65)+n)%26
        C=C+chr(ordre+65)
    return C
```
