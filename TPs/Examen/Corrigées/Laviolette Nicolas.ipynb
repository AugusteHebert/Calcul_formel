```python
#Qu1
alpha="abcdefghijklmnopqrstuvwxyz"
def indice(l):
    i=0
    while alpha[i]!=l:
        i+=1
    return i

def chiffr (cle,texte):
    res=[]
    for i in range (len(texte)):
        res.append(   alpha[  (indice(texte[i])  +   indice(  cle[i%(len(cle))])  )%26  ]    )
    return res
```

```python
#Qu2
def dechiffr (cle,texte):
    res=[]
    for i in range (len(texte)):
        res.append(   alpha[  (indice(texte[i])  -   indice(  cle[i%(len(cle))])  )%26  ]    )
    return res
```
