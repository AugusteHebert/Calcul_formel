```python
## question 3:

def Dirichlet(m, n, K):
    assert gcd(m, n)==1, " m et n pas premiers entre eux"
    
    primes = []
    current_prime = 2
    while len(primes) < K :
        if current_prime % n == m : primes.append(current_prime)
        current_prime = next_prime(current_prime)
        
    return primes
```

```python
## quetion 4:  

def triplet(c_min:int, c_max:int):
    ''' renvoie la liste des (a,b,c) tels que a**2+b**2 = c**2, pour c parcourant [[ c_min, c_max [[ et a,b des entiers naturels non nuls'''
   
    triplet_list = []

    for a in range(1, c_max):                 ## lorsque a>c_max alors a**2+b**2 > c_max**2 > c
        for b in range(1, c_max):             ## idem
            for c in range(c_min, c_max):
                if a**2+b**2 == c**2 :
                    triplet_list.append((a,b,c))
   
    return triplet_list

print(triplet(0, 26))
```

```python
## question 1:

def u(n):
    ''' méthode naive : trop appel recursif'''
    if n==0: return 0
    if n==1: return 1
    return u(n)+u(n-1)

def u(n):
    ''' méthode plus mieux : pas pafait gestion entiers trop grand'''
    u0, u1 = 0, 1
    for _ in range(n):
        u0, u1 = u1, u0+u1
    return u0

def u_mod(n, p):
    ''' méthode encore plus mieux : réduit mod pa chaque étape'''
    u0, u1 = 0, 1
    for _ in range(n):
        u0, u1 = u1, (u0+u1)%p
    return u0
```

```python
## question 5:

def break_RSA(C, n, e):
    facs = factor(n)
    p, q = facs[0][0], facs[1][0] # n=pq avec p et q premiers distincts donc factors(n) renvoie assurément 'p * q' : on a bien p=facs[0][0] et q=facs[1][0]
    
    phi = (p-1)*(q-1)
    _, d, _ = xgcd(e, phi)
    
    m = power_mod(C, d, phi)
    return m
```
